```python
import numpy as np
from metaquack import initializer as ini
from metaquack.pipeline import Evaluate, Initialize, Operator
from metaquack.lib.ga.mutation import GaussianMutation
from metaquack.lib.ga.selection import SelectBest
from metaquack.lib.ga.crossover import TwoPointCrossover
```

```python
def sphere(x):
    return np.sum(x**2, axis=1)

fields = [
    ini.Bound(-5, 5),
    ini.Bound(-5, 5),
    ini.Bound(-5, 5),
]

algo = (
    Initialize(fields, count=50)
    >> Evaluate(sphere) # fitness
    >> SelectBest(25)
    >> TwoPointCrossover()
    >> GaussianMutation(0.5)
).repeat(100)

final_state = algo.run()
```

```python
final_state.population[np.argmax(final_state.objective)]
```
